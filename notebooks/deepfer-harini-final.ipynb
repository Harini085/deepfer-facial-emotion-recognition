### Dataset Loading

```python
import os

# ==========================
# Local Dataset Paths (VS Code)
# ==========================
BASE_PATH = "archive-3"

TRAIN_DIR = os.path.join(BASE_PATH, "train")
TEST_DIR = os.path.join(BASE_PATH, "test")

# ==========================
# Dataset Validation
# ==========================
def validate_dataset_paths(train_path, test_path):
    """
    Validates the existence of dataset paths and subdirectories.
    Prevents runtime errors before loading heavy data structures.
    """

    if not os.path.exists(train_path):
        raise FileNotFoundError(
            f"❌ Training directory not found: {train_path}"
        )

    if not os.path.exists(test_path):
        raise FileNotFoundError(
            f"❌ Testing directory not found: {test_path}"
        )

    train_classes = sorted([
        d for d in os.listdir(train_path)
        if os.path.isdir(os.path.join(train_path, d))
    ])

    test_classes = sorted([
        d for d in os.listdir(test_path)
        if os.path.isdir(os.path.join(test_path, d))
    ])

    if train_classes != test_classes:
        print("⚠️ Warning: Train and Test folders have different classes.")
        print("Train:", train_classes)
        print("Test :", test_classes)
    else:
        print("✅ Dataset loaded successfully!")
        print(f"Found {len(train_classes)} emotion classes:")
        print(train_classes)

# Validate dataset
validate_dataset_paths(TRAIN_DIR, TEST_DIR)
```

```python
# For an image dataset, 'rows' correspond to total image samples and 'columns' correspond to image spatial properties
def get_dataset_summary(train_dir, test_dir):
    data = []
    classes = sorted([d for d in os.listdir(train_dir) if os.path.isdir(os.path.join(train_dir, d))])

    for emotion in classes:
        train_count = len(glob.glob(os.path.join(train_dir, emotion, '*.*')))
        test_count = len(glob.glob(os.path.join(test_dir, emotion, '*.*')))
        total_count = train_count + test_count
        data.append({
            'Emotion Class': emotion,
            'Train Images': train_count,
            'Test Images': test_count,
            'Total Images': total_count,
            'Train Ratio (%)': round((train_count / total_count) * 100, 2) if total_count > 0 else 0
        })

    df_summary = pd.DataFrame(data)
    return df_summary

df_counts = get_dataset_summary(TRAIN_DIR, TEST_DIR)
print("=== DATASET SAMPLE COUNT SUMMARY ===")
display(df_counts)

print(f"\nTotal Training Images: {df_counts['Train Images'].sum():,}")
print(f"Total Testing Images: {df_counts['Test Images'].sum():,}")
print(f"Grand Total Images: {df_counts['Total Images'].sum():,}")
```

#### Duplicate & Corrupt File Quality Checks

```python
# Check for missing/corrupt image files and duplicate image hashes
def check_image_integrity_and_duplicates(base_dir):
    corrupt_files = []
    hashes = {}
    duplicates = []

    all_files = glob.glob(os.path.join(base_dir, '**', '*.*'), recursive=True)
    image_files = [f for f in all_files if f.lower().endswith(('.png', '.jpg', '.jpeg', '.bmp'))]

    for img_path in image_files:
        # 1. Check for corrupt image files
        try:
            with Image.open(img_path) as img:
                img.verify() # Verify file integrity
        except Exception:
            corrupt_files.append(img_path)
            continue

        # 2. Check for duplicate image content using MD5 hashing
        try:
            with open(img_path, 'rb') as f:
                file_hash = hashlib.md5(f.read()).hexdigest()
                if file_hash in hashes:
                    duplicates.append((img_path, hashes[file_hash]))
                else:
                    hashes[file_hash] = img_path
        except Exception as e:
            print(f"Error hashing file {img_path}: {e}")

    return corrupt_files, duplicates

print("Scanning training dataset for quality issues...")
corrupt_train, dupes_train = check_image_integrity_and_duplicates(TRAIN_DIR)

print(f"\n=== QUALITY CHECK RESULTS ===")
print(f"Corrupt or Unreadable Images Found: {len(corrupt_train)}")
print(f"Duplicate Image Content Found: {len(dupes_train)}")

if len(corrupt_train) > 0:
    print(f"Corrupt Files: {corrupt_train}")
if len(dupes_train) > 0:
    print(f"Sample Duplicates (First 3): {dupes_train[:3]}")
```

*Note: For an image dataset, we first extract statistical features (brightness, contrast, edge intensity) to perform tabular EDA alongside spatial visual analysis.*

```python
import cv2
from tqdm import tqdm

def extract_image_features(directory):
    features = []
    classes = sorted([d for d in os.listdir(directory) if os.path.isdir(os.path.join(directory, d))])

    for emotion in classes:
        folder_path = os.path.join(directory, emotion)
        image_files = glob.glob(os.path.join(folder_path, '*.*'))

        # Sample up to 1000 images per class for faster EDA
        for img_path in tqdm(image_files[:1000], desc=f"Processing {emotion}"):
            img = cv2.imread(img_path, cv2.IMREAD_GRAYSCALE)
            if img is not None:
                brightness = np.mean(img)
                contrast = np.std(img)
                edges = cv2.Canny(img, 100, 200)
                edge_intensity = np.mean(edges) / 255.0

                features.append({
                    'Emotion': emotion,
                    'Brightness': brightness,
                    'Contrast': contrast,
                    'Edge_Intensity': edge_intensity,
                    'Width': img.shape[1],
                    'Height': img.shape[0]
                })
    return pd.DataFrame(features)

print("Extracting EDA features from training set...")
eda_df = extract_image_features(TRAIN_DIR)
print("✅ Feature extraction complete.")
```

## ***5. Hypothesis Testing***

### Based on your chart experiments, define three hypothetical statements from the dataset. In the next three questions, perform hypothesis testing to obtain final conclusion about the statements through your code and statistical testing.

*This section formulates three testable hypotheses grounded in the visual patterns observed during EDA, and validates each one using an appropriate statistical test.*

---

```python
# ==========================================
# Hypothesis Testing Setup
# ==========================================

from scipy import stats
import numpy as np

# Significance level
alpha = 0.05

print("Hypothesis testing environment is ready.")
print(f"Significance Level (α): {alpha}")
```

### Hypothetical Statement - 1
**Hypothesis:** The average brightness of facial images differs between the Happy and Sad emotion classes.

#### 1. State Your research hypothesis as a null hypothesis and alternate hypothesis.
- **Null Hypothesis (H₀):**
There is no significant difference in the mean brightness of facial images between the Happy and Sad emotion classes.

- **Alternative Hypothesis (H₁):**
There is a significant difference in the mean brightness of facial images between the Happy and Sad emotion classes.

#### 2. Perform an appropriate statistical test.

```python
from scipy import stats

# Filter brightness values
happy_brightness = eda_df[
    eda_df['Emotion'] == 'happy'
]['Brightness'].dropna()

sad_brightness = eda_df[
    eda_df['Emotion'] == 'sad'
]['Brightness'].dropna()

# Welch's Independent t-test
t_stat, p_value = stats.ttest_ind(
    happy_brightness,
    sad_brightness,
    equal_var=False
)

print(f"T-statistic: {t_stat:.4f}")
print(f"P-value: {p_value:.6f}")
```

##### Which statistical test have you done to obtain P-Value?

Welch's Independent Two-Sample t-test was performed to compare the mean brightness of the Happy and Sad emotion classes.

##### Why did you choose the specific statistical test?

Welch's Independent Two-Sample t-test was selected because it compares the means of two independent groups without assuming that they have equal variances. Since the analysis aims to determine whether the mean brightness differs significantly between the Happy and Sad emotion classes, this test provides a reliable and statistically appropriate method for hypothesis testing.

### Hypothetical Statement - 2
**Hypothesis:** The mean Edge Intensity differs across the seven emotion classes.

#### 1. State Your research hypothesis as a null hypothesis and alternate hypothesis.
*   **Null Hypothesis ($H_0$):** There is no significant difference in the mean Edge Intensity among the seven emotion classes.

*   **Alternate Hypothesis ($H_1$):** 
There is a significant difference in the mean Edge Intensity among the seven emotion classes.

#### 2. Perform an appropriate statistical test.

```python
# Group Edge Intensity values by emotion class
emotion_groups = [
    group['Edge_Intensity'].dropna().values
    for _, group in eda_df.groupby('Emotion')
]

# Perform One-Way ANOVA
f_stat, p_value = stats.f_oneway(*emotion_groups)

print(f"F-statistic: {f_stat:.4f}")
print(f"P-value: {p_value:.6f}")

# Decision based on significance level
if p_value < alpha:
    print("Conclusion: Reject the Null Hypothesis.")
    print("Edge Intensity differs significantly among the emotion classes.")
else:
    print("Conclusion: Fail to reject the Null Hypothesis.")
    print("No significant difference in Edge Intensity among the emotion classes.")
```

##### Which statistical test have you done to obtain P-Value?
One-Way Analysis of Variance (ANOVA) was performed to compare the mean Edge Intensity across the seven emotion classes and determine whether any statistically significant differences exist among them.

##### Why did you choose the specific statistical test?
One-Way ANOVA was selected because it is designed to compare the means of more than two independent groups. Since this analysis involves comparing the mean Edge Intensity across seven different emotion classes, ANOVA is the most appropriate statistical test for determining whether the observed differences are statistically significant.

### Hypothetical Statement - 3
**Hypothesis:** There is a significant linear relationship between Brightness and Contrast in facial images.

#### 1. State Your research hypothesis as a null hypothesis and alternate hypothesis.
*   **Null Hypothesis ($H_0$):** There is no significant linear relationship between Brightness and Contrast in facial images.
*   **Alternate Hypothesis ($H_1$):** There is a significant linear relationship between Brightness and Contrast in facial images.

#### 2. Perform an appropriate statistical test.

```python
# Perform Pearson Correlation Test

corr_coef, p_value = stats.pearsonr(
    eda_df['Brightness'],
    eda_df['Contrast']
)

print(f"Correlation Coefficient: {corr_coef:.4f}")
print(f"P-value: {p_value:.6f}")

# Decision based on significance level
if p_value < alpha:
    print("Conclusion: Reject the Null Hypothesis.")
    print("A significant linear relationship exists between Brightness and Contrast.")
else:
    print("Conclusion: Fail to reject the Null Hypothesis.")
    print("No significant linear relationship exists between Brightness and Contrast.")
```

##### Which statistical test have you done to obtain P-Value?
Pearson Correlation Test was performed to measure the strength and statistical significance of the linear relationship between Brightness and Contrast.

##### Why did you choose the specific statistical test?
The Pearson Correlation Test was selected because it is designed to measure the strength and direction of the linear relationship between two continuous numerical variables. Since Brightness and Contrast are continuous image features, this test is appropriate for determining whether a statistically significant linear relationship exists between them.

```python
import torch
import torch.nn as nn
from torchvision.models import convnext_tiny, ConvNeXt_Tiny_Weights

# ============================================================
# Device Configuration
# ============================================================

if torch.backends.mps.is_available():
    device = torch.device("mps")
elif torch.cuda.is_available():
    device = torch.device("cuda")
else:
    device = torch.device("cpu")

print("=" * 60)
print(f"Using Device : {device}")

if device.type == "cuda":
    print(f"NVIDIA GPU   : {torch.cuda.get_device_name(0)}")
elif device.type == "mps":
    print("Apple GPU    : Metal Performance Shaders (MPS)")
else:
    print("Running on CPU")

print("=" * 60)

# ============================================================
# Model 1 : ConvNeXt-Tiny
# ============================================================

weights = ConvNeXt_Tiny_Weights.IMAGENET1K_V1

model_1 = convnext_tiny(weights=weights)

print("✅ Pretrained ConvNeXt-Tiny Loaded")

# ============================================================
# Freeze Feature Extractor
# ============================================================

for param in model_1.features.parameters():
    param.requires_grad = False

# ============================================================
# Unfreeze Last Stage
# ============================================================

for param in model_1.features[-1].parameters():
    param.requires_grad = True

print("✅ Last ConvNeXt Stage Unfrozen")

# ============================================================
# Replace Classifier
# ============================================================

in_features = model_1.classifier[2].in_features

model_1.classifier = nn.Sequential(
    nn.Flatten(),
    nn.LayerNorm(in_features),
    nn.Dropout(0.40),
    nn.Linear(in_features, 512),
    nn.GELU(),
    nn.Dropout(0.30),
    nn.Linear(512, 7)
)

# ============================================================
# Xavier Initialization
# ============================================================

def initialize_weights(module):
    if isinstance(module, nn.Linear):
        nn.init.xavier_uniform_(module.weight)
        if module.bias is not None:
            nn.init.zeros_(module.bias)

model_1.classifier.apply(initialize_weights)

# ============================================================
# Move Model to Device
# ============================================================

model_1 = model_1.to(device)

# ============================================================
# Parameter Count
# ============================================================

trainable_params = sum(
    p.numel()
    for p in model_1.parameters()
    if p.requires_grad
)

total_params = sum(
    p.numel()
    for p in model_1.parameters()
)

print("=" * 70)
print("🚀 ConvNeXt-Tiny Ready")
print(f"Total Parameters      : {total_params:,}")
print(f"Trainable Parameters  : {trainable_params:,}")
print("=" * 70)
```

### 6. Model Training — Key Concepts Applied

Before reviewing the training loop and its output curves, it is useful to define the core training concepts referenced throughout this notebook:

- **Overfitting:** occurs when a model fits the training data (including its noise) too closely, resulting in a large gap between training and validation performance (high training accuracy, materially lower validation accuracy). Watch the Training-vs-Validation curves in Section 7 for this pattern.
- **Underfitting:** occurs when a model is too simple, undertrained, or over-regularized to capture the underlying patterns, resulting in poor performance on *both* training and validation data.
- **Regularization:** any technique that discourages the model from fitting noise, trading a small amount of training-set fit for better generalization. Here this includes weight decay (AdamW), label smoothing, and data augmentation.
- **Dropout:** randomly deactivates a fraction of neurons in the classifier head during training, forcing the network to not rely on any single neuron/feature combination and thereby improving generalization.
- **Early Stopping:** monitors validation loss/accuracy during training and halts training once it stops improving for a set number of epochs (patience), restoring the best-performing checkpoint. This both prevents overfitting and saves compute.
- **Learning Rate Scheduling (`ReduceLROnPlateau`):** automatically lowers the learning rate when validation performance plateaus, allowing the optimizer to take smaller, more precise steps as training converges.
- **Gradient Clipping:** caps the magnitude of gradients during backpropagation, preventing unstable/exploding updates — particularly useful when fine-tuning large pretrained models with a mix of frozen and trainable layers.

The training loop, executed below, applies all of these concepts consistently across all three models to keep the comparison in Section 8 fair.

```python
import os
import copy
import torch

# ============================================================
# Create Checkpoint Directory
# ============================================================

os.makedirs("checkpoints", exist_ok=True)

CHECKPOINT_PATH_1 = "checkpoints/convnext_tiny_checkpoint.pth"
BEST_MODEL_PATH_1 = "checkpoints/best_convnext_tiny.pth"

# ============================================================
# Training Configuration
# ============================================================

EPOCHS_1 = 35
start_epoch_1 = 0

train_losses_1 = []
val_losses_1 = []

train_accs_1 = []
val_accs_1 = []

best_val_acc = 0.0
best_val_loss = float("inf")

best_model_wts = copy.deepcopy(model_1.state_dict())

patience = 8
patience_counter = 0

print("\n" + "=" * 70)
print(f"🚀 Training ConvNeXt-Tiny on {device}")
print("=" * 70)

# ============================================================
# Training Loop
# ============================================================

for epoch in range(start_epoch_1, EPOCHS_1):

    # ========================================================
    # Training Phase
    # ========================================================

    model_1.train()

    running_loss = 0.0
    running_correct = 0
    total = 0

    for inputs, labels in train_loader:

        inputs = inputs.to(device)
        labels = labels.to(device)

        optimizer_1.zero_grad()

        outputs = model_1(inputs)

        loss = criterion(outputs, labels)

        loss.backward()

        # ====================================================
        # Gradient Clipping
        # ====================================================

        torch.nn.utils.clip_grad_norm_(
            filter(lambda p: p.requires_grad, model_1.parameters()),
            max_norm=1.0
        )

        optimizer_1.step()

        running_loss += loss.item() * inputs.size(0)

        _, preds = torch.max(outputs, 1)

        running_correct += (preds == labels).sum().item()

        total += labels.size(0)

    epoch_train_loss = running_loss / len(train_loader.dataset)
    epoch_train_acc = running_correct / total

    train_losses_1.append(epoch_train_loss)
    train_accs_1.append(epoch_train_acc)

    # ========================================================
    # Validation Phase
    # ========================================================

    model_1.eval()

    val_running_loss = 0.0
    val_running_correct = 0
    val_total = 0

    with torch.no_grad():

        for inputs, labels in val_loader:

            inputs = inputs.to(device)
            labels = labels.to(device)

            outputs = model_1(inputs)

            loss = criterion(outputs, labels)

            val_running_loss += loss.item() * inputs.size(0)

            _, preds = torch.max(outputs, 1)

            val_running_correct += (preds == labels).sum().item()

            val_total += labels.size(0)

    epoch_val_loss = val_running_loss / len(val_loader.dataset)
    epoch_val_acc = val_running_correct / val_total

    val_losses_1.append(epoch_val_loss)
    val_accs_1.append(epoch_val_acc)

    # ========================================================
    # Learning Rate Scheduler
    # (Monitor Validation Loss)
    # ========================================================

    scheduler_1.step(epoch_val_loss)

    current_lr = optimizer_1.param_groups[0]["lr"]

    # ========================================================
    # Save Best Model
    # Priority: Validation Accuracy
    # Tie-breaker: Validation Loss
    # ========================================================

    if (
        epoch_val_acc > best_val_acc
        or (
            abs(epoch_val_acc - best_val_acc) < 1e-6
            and epoch_val_loss < best_val_loss
        )
    ):

        best_val_acc = epoch_val_acc
        best_val_loss = epoch_val_loss

        best_model_wts = copy.deepcopy(model_1.state_dict())

        torch.save(
            {
                "epoch": epoch,
                "model_state_dict": model_1.state_dict(),
                "optimizer_state_dict": optimizer_1.state_dict(),
                "scheduler_state_dict": scheduler_1.state_dict(),
                "best_val_acc": best_val_acc,
                "best_val_loss": best_val_loss,
            },
            BEST_MODEL_PATH_1,
        )

        patience_counter = 0

        print("🏆 Best ConvNeXt-Tiny Model Updated!")

    else:

        patience_counter += 1

        print(f"⏳ Early Stopping Counter: {patience_counter}/{patience}")

    # ========================================================
    # Save Checkpoint
    # ========================================================

    torch.save(
        {
            "epoch": epoch,
            "model_state_dict": model_1.state_dict(),
            "optimizer_state_dict": optimizer_1.state_dict(),
            "scheduler_state_dict": scheduler_1.state_dict(),
            "train_loss": epoch_train_loss,
            "val_loss": epoch_val_loss,
            "train_acc": epoch_train_acc,
            "val_acc": epoch_val_acc,
            "best_val_acc": best_val_acc,
            "best_val_loss": best_val_loss,
        },
        CHECKPOINT_PATH_1,
    )

    # ========================================================
    # Epoch Summary
    # ========================================================

    print(
        f"\nEpoch [{epoch + 1}/{EPOCHS_1}]"
        f"\nTrain Loss      : {epoch_train_loss:.4f}"
        f"\nTrain Accuracy  : {epoch_train_acc:.4f}"
        f"\nValidation Loss : {epoch_val_loss:.4f}"
        f"\nValidation Acc  : {epoch_val_acc:.4f}"
        f"\nLearning Rate   : {current_lr:.6f}"
    )

    print("-" * 70)

    # ========================================================
    # Early Stopping
    # ========================================================

    if patience_counter >= patience:

        print("\n🛑 Early Stopping Triggered!")
        break

# ============================================================
# Load Best Model
# ============================================================

model_1.load_state_dict(best_model_wts)

print("\n" + "=" * 70)
print("✅ ConvNeXt-Tiny Training Completed Successfully")
print(f"🏆 Best Validation Accuracy : {best_val_acc:.4f}")
print(f"📉 Best Validation Loss     : {best_val_loss:.4f}")
print("=" * 70)
```

#### 7. Evaluation Metrics — What Each One Means

The evaluation code below computes and visualizes several standard classification metrics. Definitions, for reference throughout Sections 7–9:

| Metric | Formula (informal) | What it tells us |
|---|---|---|
| **Accuracy** | Correct predictions / Total predictions | Overall correctness; can be misleading under class imbalance |
| **Precision** | TP / (TP + FP), per class then averaged | Of the images predicted as class *X*, how many actually were *X*? High precision = few false alarms |
| **Recall** | TP / (TP + FN), per class then averaged | Of all true *X* images, how many did the model correctly find? High recall = few missed cases |
| **F1-Score** | Harmonic mean of Precision & Recall | A single balanced score, useful when both false positives and false negatives matter |
| **Macro F1** | Unweighted mean of per-class F1 | Treats every class equally regardless of size — important given FER2013's imbalance |
| **Weighted F1** | F1 averaged, weighted by class support | Reflects real-world class frequency while still accounting for per-class performance |
| **Confusion Matrix** | Cross-tabulation of predicted vs. true labels | Reveals *which* classes are most often confused with each other (e.g., Fear vs. Surprise) |

Training-vs-validation accuracy/loss curves (plotted below) are the primary diagnostic for overfitting/underfitting described in Section 6.

```python
import torch
import torch.nn as nn

from torchvision.models import (
    resnet18,
    ResNet18_Weights
)

# ============================================================
# Load Pretrained ResNet18
# ============================================================

weights = ResNet18_Weights.IMAGENET1K_V1
model_2 = resnet18(weights=weights)

print("✅ Pretrained ResNet18 Loaded")

# ============================================================
# Freeze Feature Extraction Layers
# ============================================================

for param in model_2.parameters():
    param.requires_grad = False

# ============================================================
# Unfreeze Last Residual Block (Layer4)
# ============================================================

for param in model_2.layer4.parameters():
    param.requires_grad = True

print("✅ Last ResNet Block Unfrozen")

# ============================================================
# Custom Classifier
# ============================================================

in_features = model_2.fc.in_features

model_2.fc = nn.Sequential(

    nn.Dropout(0.40),

    nn.Linear(in_features, 512),

    nn.GELU(),

    nn.Dropout(0.30),

    nn.Linear(512, 7)

)

# ============================================================
# Xavier Initialization
# ============================================================

for m in model_2.fc:

    if isinstance(m, nn.Linear):

        nn.init.xavier_uniform_(m.weight)
        nn.init.zeros_(m.bias)

# ============================================================
# Move Model to Device
# ============================================================

device = torch.device(
    "mps"
    if torch.backends.mps.is_available()
    else "cuda"
    if torch.cuda.is_available()
    else "cpu"
)

model_2 = model_2.to(device)

print(f"\nUsing Device : {device}")

# ============================================================
# Trainable Parameters
# ============================================================

total_params = sum(
    p.numel() for p in model_2.parameters()
)

trainable_params = sum(
    p.numel()
    for p in model_2.parameters()
    if p.requires_grad
)

print("\n" + "=" * 70)
print("📊 ResNet18 Model Summary")
print("=" * 70)
print(f"Total Parameters      : {total_params:,}")
print(f"Trainable Parameters  : {trainable_params:,}")
print("=" * 70)
```

```python
# ============================================================
# MODEL 3 : SWIN V2 TINY FOR FACIAL EMOTION RECOGNITION
# ============================================================

import torch
import torch.nn as nn

from torchvision.models import (
    swin_v2_t,
    Swin_V2_T_Weights
)

# ============================================================
# Load Pretrained Swin V2 Tiny
# ============================================================

weights = Swin_V2_T_Weights.IMAGENET1K_V1
model_3 = swin_v2_t(weights=weights)

print("✅ Pretrained Swin V2 Tiny Loaded")

# ============================================================
# Freeze Backbone
# ============================================================

for param in model_3.parameters():
    param.requires_grad = False

# ============================================================
# Replace Classification Head
# ============================================================

in_features = model_3.head.in_features

model_3.head = nn.Sequential(
    nn.Dropout(0.50),
    nn.Linear(in_features, 512),
    nn.GELU(),
    nn.Dropout(0.30),
    nn.Linear(512, 7)
)

# ============================================================
# Xavier Initialization
# ============================================================

for m in model_3.head:
    if isinstance(m, nn.Linear):
        nn.init.xavier_uniform_(m.weight)
        nn.init.zeros_(m.bias)

# ============================================================
# Unfreeze Last Swin Stage
# ============================================================

for param in model_3.features[-1].parameters():
    param.requires_grad = True

# ============================================================
# Ensure Head is Trainable
# ============================================================

for param in model_3.head.parameters():
    param.requires_grad = True

print("✅ Last Swin Stage + Classification Head Unfrozen")

# ============================================================
# Move Model to Device
# ============================================================

device = torch.device(
    "mps"
    if torch.backends.mps.is_available()
    else "cuda"
    if torch.cuda.is_available()
    else "cpu"
)

model_3 = model_3.to(device)

print(f"\nUsing Device : {device}")

# ============================================================
# Count Parameters
# ============================================================

total_params = sum(p.numel() for p in model_3.parameters())

trainable_params = sum(
    p.numel()
    for p in model_3.parameters()
    if p.requires_grad
)

print(f"\nTotal Parameters      : {total_params:,}")
print(f"Trainable Parameters  : {trainable_params:,}")
```

## 1. Explain the ML Model used and its Performance using Evaluation Metric Score Chart

For **Model 3**, **Transfer Learning** was implemented using **Swin V2 Tiny**, a hierarchical Vision Transformer pre-trained on the **ImageNet** dataset. Unlike traditional Convolutional Neural Networks (CNNs), Swin V2 Tiny utilizes a **shifted window-based self-attention mechanism**, enabling it to efficiently capture both local and global image features while maintaining computational efficiency. This architecture is well suited for facial emotion recognition, where understanding fine-grained facial patterns and contextual relationships is essential.

Instead of training the entire network from scratch, the **pretrained backbone was frozen**, and only the **last Swin stage** along with the **custom classification head** were fine-tuned. This transfer learning strategy preserved the rich feature representations learned from ImageNet while allowing the model to adapt effectively to the FER2013 facial emotion recognition dataset. Fine-tuning only the final stage significantly reduced computational cost, minimized overfitting, and improved generalization.

The original classification head was replaced with a custom classifier consisting of a **Dropout layer (0.5)**, a **fully connected layer with 512 neurons**, **GELU activation**, another **Dropout layer (0.3)**, and a **final output layer with 7 emotion classes**. Xavier weight initialization was applied to the newly added fully connected layers to improve training stability and convergence.

The model performance was evaluated using multiple evaluation metrics, including **Accuracy, Precision, Recall, F1-score, Confusion Matrix, and Classification Report**. ## 2. Cross-Validation & Hyperparameter Tuning

The dataset was divided into separate **training, validation, and test sets** to ensure unbiased model evaluation. Hyperparameters were selected experimentally through multiple training iterations to achieve stable convergence and improved generalization performance.

The model was trained using the **AdamW** optimizer with an initial learning rate of **0.00005 (5 × 10⁻⁵)** and a **weight decay of 0.0001**. The loss function used was **CrossEntropyLoss** with **label smoothing (0.1)** to improve generalization and reduce overconfidence in predictions.

A **ReduceLROnPlateau** learning rate scheduler monitored the **validation loss** and automatically reduced the learning rate whenever the validation loss stopped improving, allowing the optimizer to converge more effectively.

To stabilize training, **gradient clipping (maximum norm = 1.0)** was applied to prevent exploding gradients. **Early Stopping** with a patience of **8 epochs** was employed to reduce overfitting by terminating training when validation performance no longer improved.

The model was trained for a maximum of **35 epochs**, and the **best-performing model was selected based on the highest validation accuracy**, with **validation loss used as a tie-breaker** when multiple epochs achieved the same validation accuracy. This strategy ensured that the final model provided the best overall generalization performance on unseen data.

```python
import os
import copy
import torch
import torch.nn as nn

# ============================================================
# Create Checkpoint Directory
# ============================================================

os.makedirs("checkpoints", exist_ok=True)

CHECKPOINT_PATH_3 = "checkpoints/swin_v2_tiny_checkpoint.pth"
BEST_MODEL_PATH_3 = "checkpoints/best_swin_v2_tiny.pth"

# ============================================================
# Training Configuration
# ============================================================

EPOCHS_3 = 35
start_epoch_3 = 0

train_losses_3 = []
val_losses_3 = []

train_accs_3 = []
val_accs_3 = []

best_val_loss = float("inf")
best_val_acc = 0.0

best_model_wts = copy.deepcopy(model_3.state_dict())

EARLY_STOPPING_PATIENCE = 8
early_stop_counter = 0

print("\n" + "=" * 70)
print(f"🚀 Fine-Tuning Swin V2 Tiny on {device}")
print("=" * 70)

# ============================================================
# Training Loop
# ============================================================

for epoch in range(start_epoch_3, EPOCHS_3):

    # ========================================================
    # Training
    # ========================================================

    model_3.train()

    running_loss = 0.0
    running_correct = 0
    total = 0

    for inputs, labels in train_loader:

        inputs = inputs.to(device)
        labels = labels.to(device)

        optimizer_3.zero_grad()

        outputs = model_3(inputs)

        loss = criterion_3(outputs, labels)

        loss.backward()

        torch.nn.utils.clip_grad_norm_(
            filter(lambda p: p.requires_grad, model_3.parameters()),
            max_norm=1.0
        )

        optimizer_3.step()

        running_loss += loss.item() * inputs.size(0)

        _, preds = torch.max(outputs, 1)

        running_correct += (preds == labels).sum().item()
        total += labels.size(0)

    epoch_train_loss = running_loss / total
    epoch_train_acc = running_correct / total

    train_losses_3.append(epoch_train_loss)
    train_accs_3.append(epoch_train_acc)

    # ========================================================
    # Validation
    # ========================================================

    model_3.eval()

    val_running_loss = 0.0
    val_running_correct = 0
    val_total = 0

    with torch.no_grad():

        for inputs, labels in val_loader:

            inputs = inputs.to(device)
            labels = labels.to(device)

            outputs = model_3(inputs)

            loss = criterion_3(outputs, labels)

            val_running_loss += loss.item() * inputs.size(0)

            _, preds = torch.max(outputs, 1)

            val_running_correct += (preds == labels).sum().item()
            val_total += labels.size(0)

    epoch_val_loss = val_running_loss / val_total
    epoch_val_acc = val_running_correct / val_total

    val_losses_3.append(epoch_val_loss)
    val_accs_3.append(epoch_val_acc)

    # ========================================================
    # Scheduler
    # ========================================================

    scheduler_3.step(epoch_val_loss)

    current_lr = optimizer_3.param_groups[0]["lr"]

    # ========================================================
    # Save Best Model (Highest Validation Accuracy)
    # ========================================================

    if (
        epoch_val_acc > best_val_acc
        or (
            abs(epoch_val_acc - best_val_acc) < 1e-6
            and epoch_val_loss < best_val_loss
        )
    ):

        best_val_acc = epoch_val_acc
        best_val_loss = epoch_val_loss

        best_model_wts = copy.deepcopy(model_3.state_dict())

        torch.save(
            {
                "epoch": epoch + 1,
                "model_state_dict": model_3.state_dict(),
                "optimizer_state_dict": optimizer_3.state_dict(),
                "scheduler_state_dict": scheduler_3.state_dict(),
                "best_val_loss": best_val_loss,
                "best_val_acc": best_val_acc,
            },
            BEST_MODEL_PATH_3,
        )

        early_stop_counter = 0

        print("🏆 Best Swin V2 Tiny Model Updated!")

    else:

        early_stop_counter += 1

        print(
            f"⏳ Early Stopping Counter: "
            f"{early_stop_counter}/{EARLY_STOPPING_PATIENCE}"
        )

    # ========================================================
    # Save Checkpoint
    # ========================================================

    torch.save(
        {
            "epoch": epoch + 1,
            "model_state_dict": model_3.state_dict(),
            "optimizer_state_dict": optimizer_3.state_dict(),
            "scheduler_state_dict": scheduler_3.state_dict(),
            "train_loss": epoch_train_loss,
            "val_loss": epoch_val_loss,
            "train_acc": epoch_train_acc,
            "val_acc": epoch_val_acc,
        },
        CHECKPOINT_PATH_3,
    )

    # ========================================================
    # Epoch Summary
    # ========================================================

    print(
        f"\nEpoch [{epoch + 1}/{EPOCHS_3}]"
        f"\nTrain Loss      : {epoch_train_loss:.4f}"
        f"\nTrain Accuracy  : {epoch_train_acc:.4f}"
        f"\nValidation Loss : {epoch_val_loss:.4f}"
        f"\nValidation Acc  : {epoch_val_acc:.4f}"
        f"\nLearning Rate   : {current_lr:.7f}"
    )

    print("-" * 70)

    # ========================================================
    # Early Stopping
    # ========================================================

    if early_stop_counter >= EARLY_STOPPING_PATIENCE:

        print("\n🛑 Early Stopping Triggered!")

        break

# ============================================================
# Restore Best Model
# ============================================================

model_3.load_state_dict(best_model_wts)

print("\n" + "=" * 70)
print("✅ Swin V2 Tiny Training Completed Successfully")
print(f"🏆 Best Validation Accuracy : {best_val_acc:.4f}")
print(f"📉 Best Validation Loss     : {best_val_loss:.4f}")
print("=" * 70)
```

## ⚖️ 11. Advantages & Limitations of the Deep Learning FER Approach Used

*Having compared the deep learning approach against classical ML and alternative architectures, this section takes a candid, balanced look at what the approach used in this project does well and where it falls short.*

### ✅ Advantages
- **High recognition accuracy** relative to classical ML baselines, as demonstrated by ConvNeXt-Tiny's 66.06% test accuracy on a genuinely difficult 7-class problem.
- **Automatic feature extraction** — no manual engineering of edge/texture descriptors required; the network learns them directly from pixels.
- **Learns complex, hierarchical facial patterns** (from low-level edges to high-level expression semantics) via successive convolutional/attention layers.
- **Robust performance across models** — all three transfer-learning models (ConvNeXt-Tiny, ResNet18, Swin V2 Tiny) achieved 60%+ test accuracy on a 7-class problem with known label noise, indicating the approach is reasonably robust to the dataset's inherent difficulty.
- **Scalable** — the same pipeline can be retrained on larger or additional emotion datasets with minimal architectural changes.
- **Adaptable** — transfer learning allows swapping backbones (as demonstrated by comparing three different architectures) with a shared training/evaluation harness.

### ⚠️ Limitations
- **Requires large datasets** to fully realize deep learning's advantage; FER2013, while usable, is modest by modern vision-dataset standards and contains acknowledged label noise.
- **Computationally expensive** to train relative to classical ML — fine-tuning three separate deep architectures requires GPU acceleration (this notebook was run with GPU acceleration, per its Colab configuration).
- **Sensitive to lighting** and image quality variation, as observed in the brightness/contrast EDA (Section 3) — extreme lighting conditions outside the training distribution may degrade performance.
- **Occlusion issues** — partial face occlusion (glasses, hands, hair, masks) is not explicitly modeled or augmented for in this pipeline, and would likely reduce real-world accuracy versus the clean, pre-cropped FER2013 test set.
- **Class-imbalance challenges** — despite weighted sampling/loss mitigation, minority classes such as *Disgust* remain harder to classify reliably, as reflected in per-class metrics within the classification reports above.
- **Hardware requirements** — real-time or edge deployment of ConvNeXt-Tiny/Swin V2 Tiny is more demanding than lightweight architectures like MobileNet (see Section 10 comparison), which is an important consideration for on-device applications.

```python
import glob
import os
import numpy as np
import matplotlib.pyplot as plt
import PIL.Image as Image
import torch
import torch.nn.functional as F


def predict_facial_emotion(image_path, model, transform, class_names):
    """
    Predict facial emotion for a single image and visualize
    the prediction with confidence scores.
    """

    if not os.path.exists(image_path):
        raise FileNotFoundError(f"Image not found: {image_path}")

    model.eval()

    # --------------------------------------------------------
    # Load Image
    # --------------------------------------------------------
    img = Image.open(image_path).convert("RGB")

    # --------------------------------------------------------
    # Preprocess
    # --------------------------------------------------------
    input_tensor = transform(img).unsqueeze(0).to(device)

    # --------------------------------------------------------
    # Prediction
    # --------------------------------------------------------
    with torch.no_grad():
        outputs = model(input_tensor)
        probabilities = F.softmax(outputs, dim=1)[0].cpu().numpy()

    predicted_idx = np.argmax(probabilities)
    predicted_class = class_names[predicted_idx]
    confidence = probabilities[predicted_idx] * 100

    print("=" * 60)
    print("FACIAL EMOTION PREDICTION")
    print("=" * 60)
    print(f"Model             : ConvNeXt-Tiny")
    print(f"Predicted Emotion : {predicted_class}")
    print(f"Confidence        : {confidence:.2f}%")
    print("=" * 60)

    # --------------------------------------------------------
    # Visualization
    # --------------------------------------------------------
    fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(12, 5))

    ax1.imshow(img)
    ax1.set_title(
        f"{predicted_class}\n({confidence:.2f}%)",
        fontsize=12,
        fontweight="bold",
        color="green",
    )
    ax1.axis("off")

    colors = ["royalblue"] * len(class_names)
    colors[predicted_idx] = "green"

    y_pos = np.arange(len(class_names))

    ax2.barh(y_pos, probabilities * 100, color=colors)
    ax2.set_yticks(y_pos)
    ax2.set_yticklabels(class_names)
    ax2.invert_yaxis()
    ax2.set_xlabel("Confidence (%)")
    ax2.set_title("Emotion Probability Distribution", fontweight="bold")

    plt.tight_layout()
    plt.show()

    return predicted_class, confidence


# ==========================================================
# Test Inference Pipeline (Final Deployment Model)
# ==========================================================

sample_images = (
    glob.glob("/content/*/*/*/*.jpg")
    + glob.glob("/content/*/*/*/*.png")
)

# Final selected model
best_model = model_1

if sample_images:

    sample_image = sample_images[0]

    print(f"Testing inference on:\n{sample_image}\n")

    predict_facial_emotion(
        sample_image,
        best_model,
        test_transform,
        class_names,
    )

else:

    print("No sample images found.")
    print(
        "Example:\n"
        "predict_facial_emotion('path_to_image.jpg', "
        "best_model, test_transform, class_names)"
    )
```

## 🧩 12. Result Analysis

*With the best model saved, reloaded, and sanity-checked on unseen images in the previous section, this section steps back to interpret **why** the results turned out the way they did.*

**Why the model performed well:** ConvNeXt-Tiny's modernized convolutional design — large-kernel depthwise convolutions, LayerNorm, and an inverted-bottleneck block structure inspired by Transformer training recipes — combined with ImageNet-pretrained weights, gave it a strong prior over natural-image features (edges, textures, shapes) that transfer well to facial structure, even though ImageNet itself contains few faces. Consistent regularization (label smoothing, weight decay, gradient clipping) and a conservative learning rate with plateau-based scheduling allowed all three models to converge stably without wild oscillation in validation performance.

**Reasons for misclassification (expected, based on EDA and known FER2013 characteristics):**
- Visually and semantically similar emotion pairs (*Fear* vs. *Surprise*; *Sad* vs. *Neutral*) are the most likely sources of confusion, consistent with the overlapping brightness/contrast/edge-intensity distributions observed in the box and violin plots (Section 3).
- FER2013 is known in the literature to contain a meaningful fraction of ambiguous or debatable human labels, which places a natural ceiling on achievable accuracy regardless of architecture.
- Low native resolution (48×48) limits the amount of fine-grained facial detail available, even after upscaling for the pretrained backbones.

**Impact of preprocessing:** Normalizing pixel intensities to match the pretrained backbones' expected input distribution, and resizing/channel-expanding images, were necessary prerequisites for transfer learning to work at all — without them the pretrained weights would receive out-of-distribution inputs and perform poorly regardless of fine-tuning.

**Effect of augmentation:** Augmentation (applied only to the training split) increases the effective diversity of training examples, which helps reduce the training/validation performance gap — a key defense against overfitting given FER2013's moderate size and class imbalance.

**Interpretation of evaluation metrics:** The consistent ranking of ConvNeXt-Tiny > Swin V2 Tiny > ResNet18 across *every* metric reported in the Section 8 benchmark table (accuracy, macro and weighted precision/recall/F1) — rather than the models trading off strengths on different metrics — indicates the ranking reflects a genuine, robust performance difference rather than a metric-specific artifact.
